```python
import numpy as np
marks = np.array([85, 90, 88, 92, 95])
print("Marks Array:")
print(marks)
```

```python
print("First Mark:", marks[0])
print("Third Mark:", marks[2])
print("Last Mark:", marks[-1])
```

```python
print("Addition (+5):", marks + 5)
print("Multiplication (*2):", marks * 2)
print("Square:", marks ** 2)
```

```python
print("Total Marks:", np.sum(marks))
print("Average Marks:", np.mean(marks))
print("Maximum Mark:", np.max(marks))
print("Minimum Mark:", np.min(marks))
```

```python
students = np.array([
    [85, 90],
    [88, 92],
    [95, 89]
])

print("2D Array:")
print(students)
```

```python
print("First Student:", students[0])
print("Second Student Second Mark:", students[1][1])
```
