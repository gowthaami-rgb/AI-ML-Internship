```python
# Day 2 - Python Basics

# Variables and Data Types
name = "Gowthaami S M"
college = "KCG College of Technology"
course = "B.Tech Artificial Intelligence and Data Science"
year = 3

print("----- Student Details -----")
print("Name:", name)
print("College:", college)
print("Course:", course)
print("Year:", year)
```

```python
# Operators
a = 20
b = 10

print("\n----- Operators -----")
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
# Loop
print("\n----- Loop -----")
print("Numbers from 1 to 5:")
for i in range(1, 6):
    print(i)
```

```python
# Function
def student_info(name, college, course):
    print("\n----- Function -----")
    print("Welcome,", name)
    print("You are studying", course)
    print("at", college)

student_info(name, college, course)

print("\nBasic Python skills developed.")
```
